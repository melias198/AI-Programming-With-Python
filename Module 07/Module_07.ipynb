<p>
Get the text from sample.txt and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_sample.txt file
</p>

```python
# open file and read
with open("sample.txt", "r") as file:
    content = file.read()
    lines = len(content.splitlines())
    words = len(content.split())
    characters = len(content.replace(" ", "").replace("\n", ""))

    print(content)
    print(lines)
    print(words)
    print(characters)
# create new file and write 
with open("counter_of_sample.txt", "w") as file:
    file.write(f"The lines count of sample.txt is {lines}\n") 
    file.write(f"The words count of sample.txt is {words}\n") 
    file.write(f"The characters count of sample.txt is {characters}") 
```

```python
# open file and read
with open("sample.txt", "r") as readfile, open("counter_of_sample.txt", "w") as writefile:
    content = readfile.read()
    lines = len(content.splitlines())
    words = len(content.split())
    characters = len(content.replace(" ", "").replace("\n", ""))

    # another way to remove nrewlines, whitespace etc
    characters = len("".join(content.split()))

    print(content)
    print(lines)
    print(words)
    print(characters)

    writefile.write(f"The lines count of sample.txt is {lines}\n") 
    writefile.write(f"The words count of sample.txt is {words}\n") 
    writefile.write(f"The characters count of sample.txt is {characters}") 

    
```

```python
from functools import reduce
# open file and read
with open("sample.txt", "r") as readfile, open("counter_of_sample.txt", "w") as writefile:
    content = readfile.readlines()
    print(content)
    lines = len(content)
    print(lines)
    
    words = list(map(lambda x: len(x.split()), content))
    print(words)
    total_words =reduce(lambda x,y : x+y, words)
    print(total_words)

    #cleaning process
    # 1. new line delete
    content = list(map(str.strip, content))
    print(content)

    # 2. space delete
    content = list(map(lambda x:x.replace(" ", ""), content))
    print(content)

    # 3. character count
    characters = list(map(lambda x:len(x), content))
    print(characters)
    total_characters = reduce(lambda x,y:x+y, characters)
    print(total_characters)

    
```

## Practice 03

<p>
    Create a text file named emails.txt containing multiple email addresses, some in uppercase and some repeated.
Write a Python program to clean the data by converting all emails to lowercase, removing duplicates, sorting them alphabetically, and saving the cleaned list to a new file named unique_emails.txt.

</p>

```python
with open("emails.txt", "r") as file, open("unique_emails.txt", "w") as writefile:
    emails = file.readlines()
    print(type(emails))
    email_sets = sorted(set(map(lambda x: x.lower(), emails)))
    print(type(email_sets))
    # print(email_sets)
    email_str = "".join(email_sets)
    print(email_str)
    
    writefile.write(email_str)
    
```

## Practice 04


<p>
Create a text file named names.txt containing a list of names.
Some names may have extra spaces, inconsistent capitalization (like “aLiCe” or “ bob ”), or appear more than once.
Write a Python program to clean the data by:

1.  Removing extra spaces.
    
2.  Converting all names so that only the first letter is capitalized.
    
3.  Removing duplicate names. 
    
4. Sorting the names alphabetically.
    
Finally, save the cleaned list into a new file named clean_names.txt.

</p>

```python
with open("names.txt", "r") as file, open("clean_names.txt", "w") as writefile:
    names = file.read()
    # print(type(names))
    
    # 1. Removing extra space
    cleaned_names = names.replace(" ", "")
    # print(type(cleaned_names))

    # 2. Converting names 
    converted_names = cleaned_names.title()
    # print(converted_names)

    # 3. Removing duplicate
    set_names = set(converted_names.split())
    # print(type(set_names))
    # print(set_names)

    # 4. Sorting Alphabetically
    sorted_names = sorted(set_names)
    # print(sorted_names)

    # Converting string from set
    result = "\n".join(sorted_names)
    # print(type(result))
    # print(result)

    writefile.write(result)
    
     
```
